```python
from agent import final_graph, EmailState
from langgraph.types import Command


config = {
    "configurable": {
        "thread_id": "test-email-1"
    }
}


# -------------------------------
# STEP 1: Start the graph
# -------------------------------

result = final_graph.invoke(
    EmailState(
        question="""
        I want to send an email requesting leave
        from September 12 2026 to September 14 2026.

        My manager's name is John
        and his email is masanisunil176@gmail.com
        """
    ),
    config
)

print("\n========== FIRST RESULT ==========")
print(result)
```

```python
# -------------------------------
# STEP 2: Approve the email
# -------------------------------

result = final_graph.invoke(
    Command(resume="yes"),
    config
)

print("\n========== AFTER APPROVAL ==========")
print(result)
```
